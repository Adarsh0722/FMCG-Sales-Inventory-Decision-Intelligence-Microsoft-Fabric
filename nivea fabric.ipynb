```sql
import pandas as pd
import re


files = {
    "Excel Export Transaction Report050626156130900.Xls": "transaction_report",
    "Product Purchase Report 060520261226.xls": "product_purchase_report",
    "Product Wise Sales Report060520261208.xls": "product_wise_sales_report",
    "Sales Analysis Report060520261245.xls": "sales_analysis_report",
    "Sales Return Report060520261211.xls": "sales_return_report",
    "Sales Value Report - Bill Wise060520261207.xls": "sales_value_bill_wise_report"
}

for file_name, table_name in files.items():

    try:
        print(f"\nProcessing: {file_name}")


        pdf = pd.read_excel(
            f"/lakehouse/default/Files/Nivea/{file_name}",
            header=2
        )

        
        pdf.columns = [
            re.sub(r'[^a-zA-Z0-9_]', '_', str(col))
            .strip('_')
            .lower()
            for col in pdf.columns
        ]

    
        pdf = pdf.loc[:, ~pdf.columns.str.startswith('unnamed')]

    
        pdf = pdf.dropna(how='all')

    
        spark_df = spark.createDataFrame(pdf)


        spark.sql(f"DROP TABLE IF EXISTS {table_name}")

        spark_df.write \
            .mode("overwrite") \
            .option("overwriteSchema", "true") \
            .saveAsTable(table_name)

        print(f"SUCCESS: {table_name}")

    except Exception as e:
        print(f"FAILED: {file_name}")
        print(str(e))

print("\nAll files processed.")
```

```sql
CREATE OR REPLACE TEMP VIEW vw_sales AS
SELECT
    doc_date,
    doc_no,
    doc_no_full,

    party_code,
    party_name,

    item_code,
    item_name,

    qty,
    free_qty,

    rate,
    mrp,

    amount,
    taxable_amount,

    CASE
        WHEN MONTH(doc_date) BETWEEN 3 AND 6 THEN 'Summer'
        WHEN MONTH(doc_date) BETWEEN 7 AND 9 THEN 'Monsoon'
        ELSE 'Winter'
    END AS season,

    YEAR(doc_date) AS sales_year,
    MONTH(doc_date) AS sales_month

FROM Nivea_LH.transaction_report;
```

```sql
SELECT

    season,

    item_code,

    item_name,

    SUM(qty) AS total_quantity,

    SUM(amount) AS total_sales,

    ROUND(AVG(rate),2) AS avg_selling_rate

FROM vw_sales

GROUP BY

    season,
    item_code,
    item_name

ORDER BY

    season,
    total_quantity DESC;
```

```sql
SELECT

    season,

    SUM(qty) AS total_qty,

    SUM(amount) AS total_sales,

    COUNT(DISTINCT item_code) AS total_products

FROM vw_sales

GROUP BY season

ORDER BY total_sales DESC;
```

```sql
WITH product_sales AS (

SELECT

    season,

    item_name,

    SUM(qty) AS total_qty,

    SUM(amount) AS total_sales

FROM vw_sales

GROUP BY

    season,
    item_name

),

ranked AS (

SELECT *,

ROW_NUMBER() OVER(

PARTITION BY season

ORDER BY total_sales DESC

) AS rn

FROM product_sales

)

SELECT *

FROM ranked

WHERE rn<=10;
```

```sql
WITH product_sales AS (
    SELECT
        item_code,
        item_name,
        SUM(amount) AS total_sales
    FROM Nivea_LH.transaction_report
    GROUP BY item_code, item_name
),

sales_rank AS (
    SELECT
        *,
        SUM(total_sales) OVER (
            ORDER BY total_sales DESC
        ) AS cumulative_sales,
        SUM(total_sales) OVER () AS overall_sales
    FROM product_sales
)

SELECT
    item_code,
    item_name,
    total_sales,
    ROUND((cumulative_sales / overall_sales) * 100, 2) AS cumulative_percentage,
    CASE
        WHEN cumulative_sales / overall_sales <= 0.80 THEN 'A'
        WHEN cumulative_sales / overall_sales <= 0.95 THEN 'B'
        ELSE 'C'
    END AS abc_category
FROM sales_rank
ORDER BY total_sales DESC;
```

```sql
WITH sales_summary AS (

SELECT
    item_code,
    item_name,
    SUM(qty) AS total_qty,
    COUNT(DISTINCT doc_date) AS selling_days,
    ROUND(SUM(qty) / COUNT(DISTINCT doc_date),2) AS avg_daily_sales

FROM Nivea_LH.transaction_report

GROUP BY
    item_code,
    item_name
)

SELECT

c.prddcode,
c.prdname,

c.saleable,

COALESCE(s.avg_daily_sales,0) AS avg_daily_sales,

CASE
    WHEN COALESCE(s.avg_daily_sales,0)=0 THEN NULL
    ELSE ROUND(c.saleable/s.avg_daily_sales,2)
END AS stock_days

FROM Nivea_LH.current_stock_report c

LEFT JOIN sales_summary s

ON c.prddcode=s.item_code

ORDER BY stock_days;
```

```sql
WITH sales_summary AS (

SELECT
    item_code,
    SUM(qty) AS total_qty,
    ROUND(SUM(qty)/COUNT(DISTINCT doc_date),2) AS avg_daily_sales

FROM Nivea_LH.transaction_report

GROUP BY item_code
)

SELECT

c.prddcode,
c.prdname,
c.saleable,

COALESCE(s.avg_daily_sales,0) AS avg_daily_sales,

CASE
    WHEN COALESCE(s.avg_daily_sales,0)=0 THEN 'Dead Stock'

    WHEN c.saleable/s.avg_daily_sales < 15 THEN 'Reorder'

    WHEN c.saleable/s.avg_daily_sales BETWEEN 15 AND 45 THEN 'Healthy'

    ELSE 'Overstock'

END AS inventory_status

FROM Nivea_LH.current_stock_report c

LEFT JOIN sales_summary s

ON c.prddcode=s.item_code;
```

```sql
WITH sales_summary AS (

SELECT
    item_code,
    SUM(qty) AS total_qty,
    ROUND(SUM(qty)/COUNT(DISTINCT doc_date),2) AS avg_daily_sales

FROM Nivea_LH.transaction_report

GROUP BY item_code
)

SELECT

c.prddcode,
c.prdname,
c.saleable,

COALESCE(s.avg_daily_sales,0) AS avg_daily_sales,

CASE
    WHEN COALESCE(s.avg_daily_sales,0)=0 THEN 'Liquidate'
    WHEN c.saleable/s.avg_daily_sales < 15 THEN 'Reorder Immediately'
    WHEN c.saleable/s.avg_daily_sales BETWEEN 15 AND 45 THEN 'Maintain Current Stock'
    ELSE 'Reduce Purchase / Run Promotion'
END AS recommended_action

FROM Nivea_LH.current_stock_report c

LEFT JOIN sales_summary s

ON c.prddcode=s.item_code;
```

```sql
WITH salesperson_sales AS (

    SELECT
        dsr_code,
        dsr_name,

        COUNT(DISTINCT bill_number) AS total_bills,

        SUM(total_pieces) AS total_pieces,

        SUM(net_amount) AS total_sales

    FROM nivea_lh.bill_wise_sales_report

    GROUP BY
        dsr_code,
        dsr_name
)

SELECT
    *,

    DENSE_RANK() OVER (
        ORDER BY total_sales DESC
    ) AS sales_rank

FROM salesperson_sales

ORDER BY
    sales_rank;
```
